## ارفع مجلد/ملف البيانات ثم اضبط المسار التالي لملف data.yaml.

```python
from ultralytics import YOLO
DATA_YAML = '/content/dataset/data.yaml'  # عدّل المسار
model = YOLO('yolov8n.pt')
model.train(data=DATA_YAML, epochs=50, imgsz=640, batch=16, name='sea_guard')
```
